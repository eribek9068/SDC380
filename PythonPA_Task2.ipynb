```python
from datetime import datetime
import pandas as pd
from IPython.display import display, Markdown

# Load the dataset into a new DataFrame
df_new = pd.read_excel('C:\\Files\\Excel\\PA_Python\\AirBnBSummary_v2_Python_Cleaned.xlsx')

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "Eribek9068"
display(Markdown(f"### {student_id} – {todays_date}")) #display Header Text

# =====================================================================
# STEP 7: Calculate and Display Requested Descriptive Statistics
# =====================================================================
total_listings_count = len(df_new) #Total count of listings
min_listing_price = df_new['price'].min() #Minimum price
max_listing_price = df_new['price'].max() #Maximum price
mean_listing_price = df_new['price'].mean() #Mean price of listings
median_reviews = df_new['number_of_reviews'].median() #Median of number_of_reviews

#Calculate Mode of minimum_nights
mode_minimum_nights = df_new['minimum_nights'].mode()[0]
std_listing_price = df_new['price'].std() #Standard Deviation of price

#Calculate Correlation between price and availability_365
correlation_price_availability = df_new['price'].corr(df_new['availability_365'])

#Display Calculated Statistics
print("Total count of listings:", total_listings_count)
print("Minimum price:", min_listing_price)
print("Maximum price:", max_listing_price)
print("Mean price of listings:", mean_listing_price)
print("Median of number_of_reviews:", median_reviews)
print("Mode of minimum_nights:", mode_minimum_nights)
print("Standard Deviation of price:", std_listing_price)
print("Correlation between price and availability_365:", correlation_price_availability)
```
